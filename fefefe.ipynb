```python
import pandas as pd
import textwrap
from IPython.display import display
import sys

# Set unlimited output
pd.set_option('display.max_rows', None)
pd.set_option('display.max_colwidth', None)
sys.stdout.flush()

csv_path = "./candidates.csv"
df = pd.read_csv(csv_path)

# Display propuestas with text wrapping to avoid horizontal scrolling
for idx, row in df.iterrows():
    print(f"\n{'='*70}")
    print(f"Row {idx}: {row.get('nombre', 'N/A')}")
    print(f"{'='*70}")
    propuesta_text = str(row['propuestas'])
    wrapped = textwrap.fill(propuesta_text, width=70)
    print(wrapped)
    print()
```
